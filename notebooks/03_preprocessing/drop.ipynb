```python
import pandas as pd
from pathlib import Path

# ==============================
# Configuration
# ==============================

INPUT_FILE = "risk.csv"
OUTPUT_FILE = "risk_filtered.csv"

COLUMNS_TO_DROP = [
    'race',
    'Hispanic',
    'invasive'
]

# ==============================
# Load Dataset
# ==============================

print("Loading dataset...")

df = pd.read_csv(INPUT_FILE)

original_shape = df.shape

print(f"Original dataset shape: {original_shape}")

# ==============================
# Drop Excluded Features
# ==============================

existing_columns = [
    col for col in COLUMNS_TO_DROP
    if col in df.columns
]

missing_columns = [
    col for col in COLUMNS_TO_DROP
    if col not in df.columns
]

df = df.drop(columns=existing_columns)

# ==============================
# Save Processed Dataset
# ==============================

df.to_csv(OUTPUT_FILE, index=False)

# ==============================
# Validation & Summary
# ==============================

print("\nProcessing completed successfully.")
print(f"Dropped columns: {existing_columns}")
print(f"Columns not found: {missing_columns}")
print(f"Original shape: {original_shape}")
print(f"Final shape: {df.shape}")
print(f"Output file: {Path(OUTPUT_FILE).resolve()}")

# Verify that excluded columns are no longer present
assert not any(col in df.columns for col in COLUMNS_TO_DROP)

print("\nValidation passed: all excluded columns were removed.")
```
